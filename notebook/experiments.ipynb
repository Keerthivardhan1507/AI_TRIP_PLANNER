```python
from langchain_core.tools import StructuredTool
```

```python
from pydantic import BaseModel,Field
```

```python
class Weather(BaseModel):
    city:str
```

```python
def get_weather(city : str)-> str:
    """
    Get the weather of the city
    
    Args:
    city : the name of the city
    
    Returns:
    str: The weather report of the city
    """
    return f"The weather in the {city} is sunny"
    
```

```python
weather_tool = StructuredTool.from_function(
    func = get_weather,
    name = "get_weather",
    description = "Featches real time weather information of city",
    args_schema = Weather
)
```

```python
class WeatherInput(BaseModel):
    city :str = Field(...,description = "city name")
    units:str = Field("metric",description = "metric or imperial")
    
class get_weather_tool(StructuredTool):
    name :ClassVar[str] = "get_weather"
    description :ClassVar[str] = (
        "Feteches Weather data from a city"
    )
    
    args_schema:ClassVar[Type[BaseModel]] = WeatherInput
    
    def _run(self,city:str,units:str = "metric")->str:
        return get_weather(city,units)
```
